```python
# Fiiling the missing values in the dataset using the mean values of their respective columns, instead of filling with random values.

df["Study_Hours"] = df["Study_Hours"].fillna(df["Study_Hours"].mean())
df["Assignment_Score"] = df["Assignment_Score"].fillna(df["Assignment_Score"].mean())
```

```python
df.isnull().sum()

# Verification of null values in the dataset.
```

```python
df.duplicated().sum()

# It counts the number of duplicate rows in the dataset.
```

```python
df = df.drop_duplicates()

# This removes the duplicate records from the dataset.
```

```python
df.duplicated().sum()

# Verification of the removal of the duplicate records from the dataset.
```
